```python
data = uproot.open("../HitsExorcist-build/test-exorcised.root:Hits")
```

```python
events = data.arrays(data.keys())
```

```python

def vis_tracks(event):
    py.offline.init_notebook_mode()
    display(HTML(
    '<script type="text/javascript" async src="https://cdnjs.cloudflare.com/ajax/libs/mathjax/2.7.1/MathJax.js?config=TeX-MML-AM_SVG"></script>'
    ))

    
    scatter = []
    
    pdgc_codes = list(set(event['pdgc'].tolist()))    
    color = px.colors.qualitative.Plotly[len(pdgc_codes) % len(px.colors.qualitative.Plotly)]
    colors = []
    lables = []
    
    for i in range(0, len(event['pdgc'].tolist())):
        pdgc = event['pdgc'].tolist()[i]
        x=event['x'].tolist()[i]
        y=event['y'].tolist()[i]
        z=event['z'].tolist()[i]
        for p in pdgc_codes:
            if pdgc == p:
                human_readable_name = get_name_from_pdgc(p)
                if p == 0:
                    human_readable_name = "Ghost"
                colors.append(pdgc_codes.index(p))
                lables.append((human_readable_name,f"({x:.2f}, {y:.2f}, {z:.2f})"))
    
    scatter.append(go.Scatter3d(x=event['x'].tolist(), y=event['y'].tolist(), z=event['z'].tolist(), text=lables,  hoverinfo='text' , mode='markers', 
        marker=dict(
        size=1,
        color=colors,
        # colorscale='Viridis',
        opacity=1)))

    process = "NC"
    if event['isCC'] and abs(event['nu_pdgc']) == 12:
        process = r"CC $\nu_e$"
    elif event['isCC'] and abs(event['nu_pdgc']) == 14:
        process = r"CC $\nu_\mu$"
    elif event['isCC'] and abs(event['nu_pdgc']) == 16:
        process = r"CC $\nu_\tau$"

    energy = r"$E_\nu$ = " + f"{event['nu_E']:.2f} GeV"

    title = f"{event['fEvent']}-{energy}"
    print(title)
    fig = go.Figure(data=scatter)
    # Customize the layout
    fig.update_layout(
     title=title,
    scene=dict(
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        zaxis_title='Z Axis'
    ),
    width=1200,
    height=800
)

    # Show the plot in the notebook
    py.offline.iplot(fig)
```
